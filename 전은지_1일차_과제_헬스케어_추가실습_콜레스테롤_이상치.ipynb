## 문제 C. 비율 계산 (D1-03 복습)

1. 흡연자(`smoking == 1`) 비율은 몇 %인가요?
2. LDL 콜레스테롤이 130 이상인 환자 비율은 몇 %인가요?


````python
# TODO

import statistics as stats

smoking_value = [patient['smoking'] for patient in patients_new]

smoking_count = sum(smoking_value)      # 흡연자 수
smoking_ratio = round( smoking_count / len(smoking_value) * 100, 1)
    # 흡연자 비율(%)

ldl_patient = [patient['ldl_cholesterol'] for patient in patients_new]
high_ldl_count = 0    # LDL >= 130인 환자 수

for ldl in ldl_patient:
    if ldl >= 130 :
        high_ldl_count += 1


high_ldl_ratio = round(high_ldl_count / len(ldl_patient) * 100, 1)# 해당 비율(%)

print("흡연자 비율:", smoking_ratio,"%")
print("고LDL 환자 비율:", high_ldl_ratio,"%")
````

````python
# 흡연자 비율
smoking_value = [patient['smoking'] for patient in patients_new]

smoking_count = sum(smoking_value)      # 흡연자 수
smoking_ratio = round(stats.mean(smoking_value) * 100, 1)

#고LDL 환자 비율
ldl_patient = [patient['ldl_cholesterol'] for patient in patients_new]

# 삼항 연산자 + 리스트 컴프리헨션 으로 이진형_명목형으로 바꿈.
# 고LDL 환자 인가요? : 0/1 으로 만드는 과정
high_ldl_flags = [1 if ldl >= 130 else 0 for ldl in ldl_patient]

high_ldl_count = sum(high_ldl_flags)
high_ldl_ratio = round(stats.mean(high_ldl_flags) * 100, 1)# 해당 비율(%)

print("흡연자 비율:", smoking_ratio,"%")
print("고LDL 환자 비율:", high_ldl_ratio,"%")
````

## 문제 E. 이상치 영향 분석 (D1-05 복습)

1. Q009 환자(LDL 245)를 **제외**하고 다시 평균과 중앙값을 계산하세요.
2. 제외 전후 평균과 중앙값의 **변화량**을 각각 구하세요.
3. 평균과 중앙값 중 어느 쪽이 더 많이 변했나요? 그 이유는 무엇일까요?


````python
ldl_patient = [patient['ldl_cholesterol'] for patient in patients_new]

ldl_without_outlier = [patient['ldl_cholesterol'] for patient in patients_new[:-1]]  # Q009(245)를 제외한 리스트의 LDL 콜레스테롤 값들

mean_no_outlier = stats.mean(ldl_without_outlier)    # TODO
median_no_outlier = stats.median(ldl_without_outlier)    # TODO


mean_change = stats.mean(ldl_patient) - mean_no_outlier # 전체 평균 - 이상치 제외 평균
median_change = stats.median(ldl_patient) - median_no_outlier # 전체 중앙값 - 이상치 제외 중앙값

print("이상치 제외 평균:", mean_no_outlier)
print("이상치 제외 중앙값:", median_no_outlier)
print("평균 변화량:", mean_change)
print("중앙값 변화량:", median_change)
````
