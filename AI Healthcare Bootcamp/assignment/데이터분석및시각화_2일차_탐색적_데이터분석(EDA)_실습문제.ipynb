# 실습과제: 당뇨병 데이터 탐색적 분석 연습

병원에서 수집한 당뇨병 예측 데이터를 활용하여 EDA(탐색적 데이터 분석) 기법을 연습해보세요.   
groupby를 활용한 그룹별 분석부터 빈도분석, 교차분석까지 의료 데이터 분석에서 필수적인 기법들을 실습합니다.

## 요구사항

1. **그룹별 분석**
   - 당뇨병 여부별 BMI, 나이의 평균, 표준편차 계산
   - 당뇨병 여부별 BMI 평균 차이 계산

2. **범주형 변수 빈도분석**
   - 성별(Sex) 분포 확인 (빈도와 비율)
   - 교육 수준(Education) 분포 분석  

3. **교차분석 (교차표)**
   - 고혈압과 당뇨병의 연관성 파악
   - 비율 기반 교차표로 패턴 확인

## 힌트

- 그룹별 분석: `df.groupby('그룹컬럼')['타겟컬럼'].agg(['mean', 'std'])`
- 빈도분석: `.value_counts()`, `.value_counts(normalize=True)` 활용
- 교차표: `pd.crosstab(df['컬럼1'], df['컬럼2'])` 사용
- 비율 교차표: `pd.crosstab(df['컬럼1'], df['컬럼2'], normalize='index')`

````python
import pandas as pd

# 당뇨병 데이터는 df 변수로 로드됨
# 주요 컬럼: Diabetes_binary, BMI, Sex, Education, HighBP, Age 등
'''
그룹별 분석

당뇨병 여부별 BMI, 나이의 평균, 표준편차 계산
당뇨병 여부별 BMI 평균 차이 계산
범주형 변수 빈도분석

성별(Sex) 분포 확인 (빈도와 비율)
교육 수준(Education) 분포 분석
교차분석 (교차표)

고혈압과 당뇨병의 연관성 파악
비율 기반 교차표로 패턴 확인
힌트
그룹별 분석: df.groupby('그룹컬럼')['타겟컬럼'].agg(['mean', 'std'])
빈도분석: .value_counts(), .value_counts(normalize=True) 활용
교차표: pd.crosstab(df['컬럼1'], df['컬럼2']) 사용
비율 교차표: pd.crosstab(df['컬럼1'], df['컬럼2'], normalize='index')
'''

# 1. 그룹별 분석
print("=== 1. 그룹별 분석 ===")
# 당뇨병 여부별 BMI, 나이의 평균과 표준편차
# 당뇨병 여부별 BMI 평균 차이 계산

diabetes_groups = df.groupby('Diabetes_binary')
diabetes_bmi_age = df.groupby('Diabetes_binary')[['BMI','Age']].agg(['mean','std'])
normal_mean = diabetes_bmi_age.loc[0, ('BMI','mean')]
diabetes_mean = diabetes_bmi_age. loc[1, ('BMI','mean')]

print(f"[당뇨병 여부별 BMI, 나이 통계] \n\n {diabetes_bmi_age}\n")
print("\n[📊그룹 간 차이 분석]")
print(f"평균 BMI 차이: {abs(normal_mean - diabetes_mean):.2f}")
print(f"💡해석: 당뇨병 환자가 정상인 환자에 비해 4.14 높은 것으로 보아, 당뇨병과 비만도가 관련이 있는 것으로 보인다.")


# 2. 범주형 변수 빈도분석
print("\n=== 2. 빈도분석 ===")
# 성별 분포 (빈도와 비율)
# 교육 수준 분포 분석

sex_counts = df['Sex'].value_counts()
sex_percent = df['Sex'].value_counts(normalize=True) *100
print("\n[성별분포]\n")
for element in sex_counts.index:
  count = sex_counts[element]
  percent = sex_percent[element]
  gender = "남성" if element == 0 else "여성"
  print(f"   {gender}: {count:,}명 ({percent:.1f}%)")


print("\n[교육 수준 분포]\n")
education_count = df['Education'].value_counts().sort_index()
for edu_level, count in education_count.items():
  None
education_percent = df['Education'].value_counts(normalize=True).sort_index() * 100
for edu_level, percent in education_percent.items():
  None

for edu_level, count, percent in zip(
    education_count.index,
    education_count.values,
    education_percent.values
):

  print(f"수준 {edu_level}: {count:,}명 ({percent:.1f}%)")

# 3. 교차분석 (교차표)
print("\n=== 3. 교차분석 ===")
# 고혈압과 당뇨병의 연관성 파악
# 비율 기반 교차표로 패턴 확인
print("\n[고혈압과 당뇨병 교차분석표]\n")
crosstab_result= pd.crosstab(df['HighBP'],df['Diabetes_binary'],margins=True)
crosstab_result.index=['정상혈압','고혈압','총계']
crosstab_result.columns=['정상','당뇨병','총계']
print(crosstab_result)


print("\n[비율 기반 교차분석표]\n")
crosstab_p = pd.crosstab(df['HighBP'], df['Diabetes_binary'], normalize = 'index') *100
crosstab_p.index = ['정상','고혈압']
crosstab_p.columns = ['정상','당뇨병']
print(round(crosstab_p,1))
````

````python
# @title 정답

import pandas as pd

# 1. 그룹별 분석
print("=== 1. 그룹별 분석 ===")

# 당뇨병 여부별 BMI와 나이의 평균, 표준편차
print("당뇨병 여부별 BMI, 나이 통계:")
diabetes_stats = df.groupby('Diabetes_binary')[['BMI', 'Age']].agg(['mean', 'std'])
diabetes_stats.index = ['정상', '당뇨병']
print(diabetes_stats.round(2))

# 당뇨병 여부별 BMI 평균 차이 계산
bmi_by_diabetes = df.groupby('Diabetes_binary')['BMI'].mean()
mean_diff = bmi_by_diabetes[1] - bmi_by_diabetes[0]  # 당뇨병 환자 - 정상인

print(f"\n📊 그룹 간 차이 분석:")
print(f"평균 BMI 차이: {mean_diff:.2f} (당뇨병 환자가 더 높음)")
print(f"💡 해석: 당뇨병 환자의 평균 BMI가 {mean_diff:.1f} 높아 비만과 당뇨병의 연관성 시사")

# 2. 범주형 변수 빈도분석
print("\n=== 2. 빈도분석 ===")

# 성별 분포 (빈도와 비율)
print("성별 분포:")
sex_counts = df['Sex'].value_counts()
sex_ratio = df['Sex'].value_counts(normalize=True)
for category in sex_counts.index:
    count = sex_counts[category]
    ratio = sex_ratio[category]
    gender = "여성" if category == 1 else "남성"
    print(f"  {gender}: {count:,}명 ({ratio:.1%})")

# 교육 수준 분포
print(f"\n교육 수준 분포:")
edu_counts = df['Education'].value_counts().sort_index()
edu_total = len(df)
for level in edu_counts.index:
    count = edu_counts[level]
    ratio = count / edu_total
    print(f"  수준 {level}: {count:,}명 ({ratio:.1%})")


# 3. 교차분석 (교차표)
print("\n=== 3. 교차분석 ===")

# 고혈압과 당뇨병의 교차표
print(f"\n고혈압 × 당뇨병 교차표 (빈도):")
bp_diabetes_crosstab = pd.crosstab(df['HighBP'], df['Diabetes_binary'], margins=True)
bp_diabetes_crosstab.index = ['정상혈압', '고혈압', '전체']
bp_diabetes_crosstab.columns = ['정상', '당뇨병', '전체']
print(bp_diabetes_crosstab)

# 고혈압과 당뇨병의 비율 교차표
print(f"\n고혈압 × 당뇨병 교차표 (비율):")
bp_diabetes_ratio = pd.crosstab(df['HighBP'], df['Diabetes_binary'], normalize='index')
bp_diabetes_ratio.index = ['정상혈압', '고혈압']
bp_diabetes_ratio.columns = ['정상', '당뇨병']
print((bp_diabetes_ratio * 100).round(1))   # 퍼센트로 표시
````
